```python
import pandas as pd
import numpy as np
import os

data_folder = r'C:\Users\Nafeesa\Desktop\IOMT\iomt-intrusion-detection\data'

csv_files = [f for f in os.listdir(data_folder) if f.endswith('.csv') and 'cleaned' not in f]

dfs = []
for file in csv_files:
    path = os.path.join(data_folder, file)
    df = pd.read_csv(path)
    attack_name = file.replace('_train.pcap.csv', '')
    df['Label'] = attack_name
    dfs.append(df)
    print(f"✅ {attack_name}: {len(df)} rows")

df = pd.concat(dfs, ignore_index=True)
print(f"\nTotal rows: {len(df)}")
print(f"Duplicates: {df.duplicated().sum()}")
print("\nClass distribution:")
print(df['Label'].value_counts())

save_path = os.path.join(data_folder, 'cleaned_data.csv')
df.to_csv(save_path, index=False)
print(f"\n✅ cleaned_data.csv save ho gaya!")
```

```python
from sklearn.preprocessing import MinMaxScaler
from sklearn.preprocessing import LabelEncoder

# Label column alag karo
X = df.drop('Label', axis=1)
y = df['Label']

# Normalization - 0 to 1 range
scaler = MinMaxScaler()
X_scaled = scaler.fit_transform(X)

print("✅ Normalization ho gaya!")
print(f"Min value: {X_scaled.min():.2f}")
print(f"Max value: {X_scaled.max():.2f}")
```

```python
from imblearn.over_sampling import SMOTE

print("Before SMOTE:")
print(df['Label'].value_counts())

smote = SMOTE(random_state=42)
X_resampled, y_resampled = smote.fit_resample(X_scaled, y)

print("\nAfter SMOTE:")
import pandas as pd
print(pd.Series(y_resampled).value_counts())
print(f"\nTotal rows after SMOTE: {len(X_resampled)}")
```

```python
from imblearn.over_sampling import SMOTE
import pandas as pd

# Sirf 10000 samples per class lo - fast hoga
df_small = df.groupby('Label').apply(lambda x: x.sample(min(len(x), 10000), random_state=42)).reset_index(drop=True)

X_small = df_small.drop('Label', axis=1)
y_small = df_small['Label']

from sklearn.preprocessing import MinMaxScaler
scaler = MinMaxScaler()
X_small_scaled = scaler.fit_transform(X_small)

smote = SMOTE(random_state=42)
X_resampled, y_resampled = smote.fit_resample(X_small_scaled, y_small)

print("After SMOTE:")
print(pd.Series(y_resampled).value_counts())
print(f"\nTotal rows: {len(X_resampled)}")
print("✅ SMOTE done!")
```

```python
import pandas as pd
import numpy as np
import os
from sklearn.preprocessing import MinMaxScaler

data_folder = r'C:\Users\Nafeesa\Desktop\IOMT\iomt-intrusion-detection\data'

df = pd.read_csv(os.path.join(data_folder, 'cleaned_data.csv'))
print(f"✅ Data loaded: {len(df)} rows")
print(df['Label'].value_counts())
```

```python
from imblearn.over_sampling import SMOTE

# Sirf 5000 samples per class - bohot fast hoga
df_small = df.groupby('Label').apply(lambda x: x.sample(min(len(x), 5000), random_state=42)).reset_index(drop=True)

X = df_small.drop('Label', axis=1)
y = df_small['Label']

scaler = MinMaxScaler()
X_scaled = scaler.fit_transform(X)

smote = SMOTE(random_state=42)
X_resampled, y_resampled = smote.fit_resample(X_scaled, y)

print("✅ SMOTE done!")
print(pd.Series(y_resampled).value_counts())
print(f"Total rows: {len(X_resampled)}")
```

```python
from imblearn.over_sampling import SMOTE

# Label column wapas add karo
df_small = df.groupby('Label').apply(lambda x: x.sample(min(len(x), 5000), random_state=42))
df_small = df_small.reset_index(level=0, drop=True)

X = df_small.drop('Label', axis=1)
y = df_small['Label']

scaler = MinMaxScaler()
X_scaled = scaler.fit_transform(X)

smote = SMOTE(random_state=42)
X_resampled, y_resampled = smote.fit_resample(X_scaled, y)

print("✅ SMOTE done!")
print(pd.Series(y_resampled).value_counts())
print(f"Total rows: {len(X_resampled)}")
```

```python
from imblearn.over_sampling import SMOTE

# Simple sampling
frames = []
for label in df['Label'].unique():
    subset = df[df['Label'] == label].sample(min(5000, len(df[df['Label'] == label])), random_state=42)
    frames.append(subset)

df_small = pd.concat(frames, ignore_index=True)

X = df_small.drop('Label', axis=1)
y = df_small['Label']

scaler = MinMaxScaler()
X_scaled = scaler.fit_transform(X)

smote = SMOTE(random_state=42)
X_resampled, y_resampled = smote.fit_resample(X_scaled, y)

print("✅ SMOTE done!")
print(pd.Series(y_resampled).value_counts())
print(f"Total rows: {len(X_resampled)}")
```

```python
import numpy as np

# Preprocessed data save karo
X_df = pd.DataFrame(X_resampled, columns=df.drop('Label', axis=1).columns)
X_df['Label'] = y_resampled.values

save_path = os.path.join(data_folder, 'preprocessed_data.csv')
X_df.to_csv(save_path, index=False)
print("✅ preprocessed_data.csv save ho gaya!")
print(f"Total rows: {len(X_df)}")
```

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder

data_folder = r'C:\Users\Nafeesa\Desktop\IOMT\iomt-intrusion-detection\data'
df = pd.read_csv(f'{data_folder}/preprocessed_data.csv')

print(f"Data loaded: {len(df)} rows")
print(df['Label'].value_counts())
```

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder

X = df.drop('Label', axis=1)
y = df['Label']

# Label encode karo
le = LabelEncoder()
y_encoded = le.fit_transform(y)

# Train/Test split
X_train, X_test, y_train, y_test = train_test_split(
    X, y_encoded, test_size=0.2, random_state=42, stratify=y_encoded)

print(f"✅ Training samples: {len(X_train)}")
print(f"✅ Testing samples: {len(X_test)}")
print(f"Classes: {le.classes_}")
```

```python
import tensorflow as tf
from tensorflow.keras.applications import MobileNetV2
from tensorflow.keras.models import Model
from tensorflow.keras.layers import Dense, GlobalAveragePooling2D, Dropout, Input, Reshape

# Data ko image shape mein convert karo
# 45 features → 5x9 image → 3 channel (MobileNetV2 needs 3 channels)
X_train_img = X_train.values.reshape(-1, 5, 9, 1)
X_train_img = np.repeat(X_train_img, 3, axis=-1)

X_test_img = X_test.values.reshape(-1, 5, 9, 1)
X_test_img = np.repeat(X_test_img, 3, axis=-1)

print(f"✅ Image shape: {X_train_img.shape}")
```
